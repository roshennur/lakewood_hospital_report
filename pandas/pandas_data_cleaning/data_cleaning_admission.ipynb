```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:admissions_m.csv')
```

```python
print(df.shape)
print(df.dtypes)
```

```python
df.info()
```

```python
# Convert columns to correct datatype

df['admission_date'] = pd.to_datetime(df['admission_date'])
df['discharge_date'] = pd.to_datetime(df['discharge_date'])

df.dtypes
```

```python
# Check for unique/missing values in column

print(df['ward_department'].value_counts())
print("\nUnique values: ", df['ward_department'].nunique())
print("\nMissing values: ", df['ward_department'].isna().sum())
```

```python
# Normalize the ward_department string column

df['ward_department'] = df['ward_department'].str.strip().str.title()
df['ward_department'] = df['ward_department'].str.replace("Icu", "ICU")
print(df['ward_department'].value_counts())
```

```python
# Check for unique and Null values in 'discharge_status'

print(df['discharge_status'].value_counts())
print("\nUnique values: ", df['discharge_status'].nunique())
print("\nMissing values: ", df['discharge_status'].isna().sum())
```

```python
# Normalize the discharge_status string column

df['discharge_status'] = df['discharge_status'].str.strip().str.title()
df['discharge_status'] = df['discharge_status'].str.replace("Recoverd", "Recovered")
df['discharge_status'] = df['discharge_status'].str.replace("Recoverred", "Recovered")
df['discharge_status'] = df['discharge_status'].str.replace("Transfered", "Transferred")
df['discharge_status'] = df['discharge_status'].str.replace("Tranferred", "Transferred")
df['discharge_status'] = df['discharge_status'].str.replace("Deceasd", "Deceased")
df['discharge_status'] = df['discharge_status'].str.replace("Deceassed", "Deceased")

print(df['discharge_status'].value_counts())
```

```python
# Create derived column lenght_of_stay

df['length_of_stay'] = (df['discharge_date'] - df['admission_date']).dt.days
df[['admission_date', 'discharge_date', 'length_of_stay']].head()
```

```python
df.to_csv('admissions_cleaned.csv', index=False)
```
